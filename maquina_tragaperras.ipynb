```python
#importamos librerías 
import random

#constantes en nuestro código
MAX_LINEAS = 3
MAX_APUESTA = 100
MIN_APUESTA = 1
FILAS= 3
COLUMNAS = 3

#Especificamos la cantidad de símbolos que va a haber en cada una

symbol_count = {
    "X": 3, 
    "Y": 4, 
    "Z": 6, 
    "W": 8 
}
symbol_value = {
    "X": 5, 
    "Y": 4, 
    "Z": 3, 
    "W": 2 
}

```

```python
def get_slot_machine_spin(filas, columnas, symbols):
    """Genera una tirada aleatoria de la máquina tragaperras.

    Args:
        filas: número de símbolos por columna.
        columnas: número de columnas.
        symbols: diccionario con cada símbolo y su cantidad disponible.

    Returns:
        list: lista de columnas con los símbolos seleccionados.
    """
    # Creamos una lista con todos los símbolos disponibles
    all_symbols = []
    for symbol, symbol_count in symbols.items():
        for _ in range(symbol_count): # Añadimos cada símbolo tantas veces como indica su cantidad
            all_symbols.append(symbol)
    # Guardamos aquí las columnas generadas
    resultado = []
    # Generamos una columna en cada repetición
    for col in range(columnas):
        columna = []
        # Copiamos la lista para que cada columna tenga todos los símbolos disponibles
        current_symbols = all_symbols[:]
        # Elegimos un símbolo para cada fila de esta columna
        for fila in range(filas):
            value = random.choice(current_symbols)
            # Quitamos una aparición del símbolo elegido de la lista disponible
            current_symbols.remove(value)
            # Añadimos el símbolo elegido a la columna
            columna.append(value)
        # Guardamos la columna completa en el resultado
        resultado.append(columna)

    return resultado
```

```python
def check_ganancias(columnas, lineas, bet, values):
    """Calcula las ganancias de una tirada y qué líneas han ganado.

    Returns:
        tuple: (dinero total ganado, lista con los números de línea ganadoras)
    """
    ganancias = 0
    # Aquí guardaremos qué líneas han ganado
    ganancias_linea = []

    for linea in range(lineas):
        symbol = columnas[0][linea]
        for columna in columnas:
            if symbol != columna[linea]:
                break
        else:
            ganancias += values[symbol] * bet
            # Sumamos 1 porque el usuario cuenta las líneas desde 1, no desde 0
            ganancias_linea.append(linea + 1)

    return ganancias, ganancias_linea
```

```python
def print_slot_machine(columnas):
    # Recorremos las filas de la máquina tragaperras.
    # len(columnas[0]) nos dice cuántos símbolos tiene la primera columna.
    for fila in range(len(columnas[0])):
        # Recorremos todas las columnas.
        for i, columna in enumerate(columnas):
            # Comprobamos si la columna actual NO es la última.
            if i != len(columnas) - 1:
                print(columna[fila], end=" | ")
            else:
                print(columna[fila], end="")
        print()
```

```python
def deposit(): 
    """Solicita una cantidad de depósito entera y mayor que cero. 

    Returns:
        int: cantidad de dinero que introduce el usuario
    """
    #mientras que el dinero sea positivo que siga pasando la condición
    while True: 
        amount = input("Cúanto dinero te gustaría depositar? $: ")
        # tenemos que comprobar si la cantidad es utilizable 
        if amount.isdigit(): 
            amount = int(amount) #lo convertimos en número 
            #tenemos que comprobar si es mayor que cero 
            if amount > 0: 
                break
            else: 
                print("El depósito tiene que ser mayor que 0$")
        else: 
            print("Por favor, introduce un número")
    return amount
deposit()#comprobamos que esta bien 
```

```python
def get_number_of_lines():
    """
    Solicita la línea de apuesta que quiere el usuario mayor que cero y dentro del rango 
    
    Returns:
        int: línea que apuesta el usuario 
    """
    #mientras que el dinero sea positivo que siga pasando la condición
    while True: 
        lineas = input("Introduce el número de líneas en las que quieres apostar (1 - " + str(MAX_LINEAS)+ ")? ")
        if lineas.isdigit(): 
            lineas = int(lineas) #lo convertimos en número 
            #tenemos que comprobar si esta en el límite que hemos puesto 
            if 1 <= lineas <= MAX_LINEAS:  
                break
            else: 
                print("Introduce un número velido de apuesta")
        else: 
            print("Por favor, introduce un número")
    return lineas
get_number_of_lines()
```

```python
def get_bet():
    """Solicita una apuesta entre los límites mínimo y máximo.

    Returns:
        int: cantidad de dinero que introduce el usuario.
    """
    # Repetimos la pregunta hasta obtener una apuesta válida
    while True:
        amount = input("¿Cuánto dinero te gustaría apostar en cada línea? $: ")
        # Comprobamos que el texto contiene solo dígitos
        if amount.isdigit():
            amount = int(amount) 
            # Comprobamos que la apuesta está dentro de los límites
            if MIN_APUESTA <= amount <= MAX_APUESTA:
                break
            else:
                print(
                    f"La apuesta tiene que estar entre ${MIN_APUESTA} y ${MAX_APUESTA}"
                )
        else:
            print("Por favor, introduce un número entero")
    return amount
```

```python
def spin(balance): 
    linea = get_number_of_lines()
    while True: 
        apuesta = get_bet()
        total_apuesta = apuesta * linea

        if total_apuesta > balance: 
            print(f"No tienes suficiente dinero para apostar esa cantidad. Tu saldo actual es de: {balance}$ ")
        else: 
            break
    print(f"Tu apuesta es ${apuesta} en la línea {linea}. Tu apuesta total es de: {total_apuesta}$ ")

    slots = get_slot_machine_spin(FILAS, COLUMNAS, symbol_count)
    print_slot_machine(slots)
    ganancias, ganancias_linea = check_ganancias(slots, linea, apuesta, symbol_value)
    print(f"Has ganado ${ganancias}")
    print(f"Has ganado en lineas:", *ganancias_linea)
    return ganancias - total_apuesta
```

```python
def main(): 
    balance = deposit()
    while True: 
        print(f"El balance actual es de ${balance}")
        respuesta = input("Presiona enter para el spin(q para quitarlo)")
        if respuesta == "q": 
            break 
        balance += spin(balance)


main()
```
